```python
import numpy as np
arr = np.array([1, 3, 5, 7])
print("numpy_array:", arr)
```

```python
arr2 = np.array([[1, 3, 5, 7], [2, 6, 3, 6]])
print("numpy_array2:", arr2)
array_shape = arr2.shape
print("array_shape:", array_shape)
```

```python
# dtype
a = np.array([1, 2, 3])
array_type = a.dtype
print("array_type:", array_type)
```

```python
# reshape
b = np.array([1, 2, 3, 4, 5, 6])
c = b.reshape(2, 3)
print("reshaped_array:", c)
```

```python
# astype()
d = np.array([1, 2, 3, 4])
e = d.astype(float)
print("float_typed:", e)
```

```python
# np.zeros(), np.ones() => dummy data
z = np.zeros((2, 3))
o = np.ones((2 ,3))
print("zeros:", z)
print("ones:", o)
```

```python
# ndim() this method tells us how many dimension our array has
a = np.array([1, 2, 3])
print("ndim:", a.ndim)
```

```python
# arange()
a = np.arange(1, 7)
print("arrange:", a)
```

```python
# sum(), mean(), max(), min()
a = np.array([1, 2, 3, 4])
print("sum:", a.sum())
print("mean:", a.mean())
print("max:", a.max())
print("min:", a.min())
```

```python
# flatten() this makes 2D, 3D arrays 1D
a = np.array([[1, 2, 3], [4, 5, 6]])
print("flatten:", a.flatten())
```

```python
# dor() or @ we use this when we wanna do mutiplication of matrix
a = np.array([[1, 3], [5, 6]])
b = np.array([[5, 4], [7, 1]])
print("dot/multipication:", a.dot(b))
```

```python
# random.rand() this creates array with random numbers
a = np.random.rand(2, 3)
print("random.rand:", a)
```
