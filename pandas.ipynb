```python
import pandas as pd
```

```python
mydataset = {
    'cars': ["BMW", "Audi", "Ford", "Tesla", "Honda", "Ferrari", "BYD"],
    'counts': [5, 7, 2, 9, 6, 7, 3]
}
df = pd.DataFrame(mydataset)
print(df)
```

```python
data_with_missing = {
    "names": ["Michael", "Herry", None, "Alex", "Sam", None, "John"],
    "ages": [23, 45, 22, None, 53, None, 32]
}
df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
df_missing.isnull().sum() # this is data cleaning "isnul" is checking if there is any data missing
```

```python
df_missing.dropna() # this method remove columns that has no data
```

```python
df_missing.fillna({"names": "unknown", "ages": 0}) # this fills missing data with given data
```
